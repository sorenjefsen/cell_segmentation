```python
import numpy as np
import matplotlib.pyplot as plt
```

Here i can play around with Rayleigh distributions and their mixtures.

```python
p = 0.5
D_f = 0.1
D_n = 0.01
dt = 10

def Rayleigh(x, sigma):
    return (x / sigma**2) * np.exp(-x**2 / (2 * sigma**2))

def mixture(x, D_f, D_n, dt, p):
    return p * Rayleigh(x, D_f*dt) + (1-p) * Rayleigh(x, D_n*dt)

x = np.linspace(0, 5, 1000)
y = mixture(x, D_f, D_n, dt, p)
plt.plot(x, y)

```
